```python
texto = "Explorando a diversidade de linguagens de programação com pyhton"

#exibindo o texto
print("="*64)
print(texto)
print("="*64)

# tamanho do texto
print(f'Tamanho do texto = {len(texto)}')

# contatando letras especificas
print(f'Quantidade de "e" na frase = {texto.count("e")}')

# pegando a primeiras palavras especificas
print(f'A 5 primeiras palavras do texto são = {texto[:5]}')

# criando umas lista

cores = ['vermelho', 'verde','amarelo','azul']

for cor in cores:
  print(f'Posição = {cores.index(cor)}, cor = {cor}')
```

```python
# tuplas - listas que não são alteradas

# usando o for com dois paramentros de posição e valor, passando dois argumentos
# como "p" e "z", onde "p" é o posição do indice e "z" é o valor da posição.
# usamos o enumerate() para fazer isso.

vogais = ('a', 'e', 'i', 'o','u')

for p,z in enumerate(vogais):
  print(f'Posição = {p}, valor = {z}')
```

```python
# lista de convidados

# fazer a lista de convidados por meio da tupla e a lista de confirmados em uma
# lista simples.

convidados = ('andre', 'jonas', 'pedro', 'maria')
confirmados = ['andre','maria']

# usando o not in para fazer a verificação da lista que foi informada do if, nesse caso vai mostrar quem não foi confirmado
# o "in" server como se pertenece ou não pertence
for convidados in convidados:
  if convidados not in confirmados:
    print(f'Ainda falta confirmar o {convidados}')
```

```python
# usando a função lambd

resultado = lambda x,y: x+y

print(f"O valor da soma de 5 + 5 é = {resultado(10,10)}")

# usando o map. Map precisa de dois parametros, 1 função e 1 interação, pode ser listas, tuplas.

def dobrar(x):
  return x + 2

number = [1,2,3,4]
resultado = map(dobrar,number)

print(list(resultado))
# funcao: Uma função que retorna True (verdadeiro) ou False (falso) para cada item. Se for None, o Python remove os valores considerados falsos (como 0, "" ou None)

def par(x):
  return x % 2 == 0

pares = [1,2,3,4,5,6,7,9,10]

result = list(filter(par,pares))
print(result)
```
